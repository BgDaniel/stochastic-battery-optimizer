```python
import pickle
from stochasticbatopt.plotting import plot_value_distribution
```

## CONFIG

```python
INSPECT_DAYS = [0, 2, 4]   # plot for each of these day indices
```

## Load model

```python
with open('../sbo_cache/model_and_result.pkl', 'rb') as fh:
    state = pickle.load(fh)

model      = state['model']
result     = state['result']
time_index = state['time_index']
print(f'Loaded model: {model.n_days} days, {model.n_levels} energy levels, '
      f'N={model.price_scenarios.shape[0]} scenarios')
```

## Value function plots

```python
%matplotlib inline
import matplotlib.pyplot as plt

for day in INSPECT_DAYS:
    plot_value_distribution(model, day=day)
    plt.show()
```

---
### Interpretation guide

| Observation | Interpretation |
|---|---|
| Steep positive slope | High marginal value of stored energy — worth charging more |
| Flat curve | Energy level at day boundary doesn't matter much |
| Narrow P5–P95 bands | Value is nearly deterministic given starting energy |
| Wide bands | Strong price-path dependency — uncertainty matters |
| Mean V ≈ intrinsic line | Extrinsic value is small (Jensen gap is small) |
